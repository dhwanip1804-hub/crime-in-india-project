```python
import matplotlib.pyplot as plt

yearly_crime = df.groupby("YEAR")["TOTAL IPC CRIMES"].sum()

plt.figure(figsize=(10,5))
plt.plot(yearly_crime.index, yearly_crime.values, marker='o')

plt.title("Total IPC Crimes Over Years (India)")
plt.xlabel("Year")
plt.ylabel("Total Crimes")
plt.grid(True)

plt.show()

```

```python
crime_columns = [
    "MURDER",
    "RAPE",
    "ROBBERY",
    "BURGLARY",
    "THEFT",
    "DOWRY DEATHS",
    "ASSAULT ON WOMEN WITH INTENT TO OUTRAGE HER MODESTY",
    "CHEATING",
    "RIOTS",
    "CAUSING DEATH BY NEGLIGENCE"
]

crime_totals = df[crime_columns].sum().sort_values(ascending=False)

plt.figure(figsize=(10,5))
crime_totals.plot(kind="bar")

plt.title("Top Crime Categories in India")
plt.ylabel("Number of Cases")
plt.xticks(rotation=45)
plt.show()
```

```python
murder_trend = df.groupby("YEAR")["MURDER"].sum()

plt.figure(figsize=(10,5))
plt.plot(murder_trend.index, murder_trend.values, color="red", marker="o")

plt.title("Murder Cases Trend Over Years")
plt.xlabel("Year")
plt.ylabel("Cases")
plt.grid(True)

plt.show()
```

```python
import pandas as pd
import os
df = pd.read_csv("C:/Users/SHREYA/Desktop/Crime-In-India-Project/dataset/raw_data/01_District_wise_crimes_committed_IPC_2001_2012.csv")
```

```python
crime_records = df.groupby("YEAR")[[
    "MURDER",
    "RAPE",
    "THEFT",
    "ROBBERY",
    "DOWRY DEATHS"
]].sum().reset_index()

crime_records.head()
```

```python
os.makedirs("../frontend/src/data", exist_ok=True)

crime_records.to_json(
    "../frontend/src/data/crime_records.json",
    orient="records"
)
```

```python
states_covered = df["STATE/UT"].unique()

states_df = pd.DataFrame(states_covered, columns=["STATE/UT"])
states_df.head()
```

```python
states_df.to_json(
    "../frontend/src/data/states_covered.json",
    orient="records"
)
```

```python
resources = [
    "District wise IPC crimes 2001-2012",
    "Crimes against SC/ST",
    "Crimes against children",
    "Juvenile crimes",
    "Women crimes dataset",
    "Human rights violations",
    "Rape victims dataset"
]

resources_df = pd.DataFrame(resources, columns=["dataset_name"])
resources_df
```

```python
resources_df.to_json(
    "../frontend/src/data/resources.json",
    orient="records"
)
```

```python
visualization = df.groupby("YEAR")[["MURDER", "RAPE", "THEFT", "ROBBERY"]].sum().reset_index()
visualization.head()
```

```python
visualization.to_json(
    "../frontend/src/data/visualization.json",
    orient="records"
)
```

```python
catalog = df[[
    "MURDER",
    "RAPE",
    "ROBBERY",
    "THEFT",
    "DOWRY DEATHS"
]].sum().reset_index()

catalog.columns = ["Crime", "Total Cases"]
catalog
```

```python
catalog.to_json(
    "../frontend/src/data/catalog.json",
    orient="records"
)
```

```python
downloads = df.groupby("YEAR")["TOTAL IPC CRIMES"].sum().reset_index()
downloads.columns = ["Year", "Total Downloads"]
downloads
```

```python
downloads.to_json(
    "../frontend/src/data/downloads.json",
    orient="records"
)
```

```python
safety = df.groupby("STATE/UT")[["MURDER", "RAPE", "THEFT"]].sum()

safety["Safety Score"] = (
    safety["MURDER"] * 0.5 +
    safety["RAPE"] * 0.3 +
    safety["THEFT"] * 0.2
)

safety = safety.reset_index()
safety.head()
```

```python
safety.to_json(
    "../frontend/src/data/safety_index.json",
    orient="records"
)
```
