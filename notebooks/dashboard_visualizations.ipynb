```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import json
```

```python
df = pd.read_csv("C:/Users/SHREYA/Downloads/India_Crime - Copy_2/India_Crime - Copy/01_District_wise_crimes_committed_IPC_2001_2012.csv"
)

df.head()
```

```python
print(df.columns)
```

```python
yearly_crime = (
    df.groupby("YEAR")["TOTAL IPC CRIMES"]
    .sum()
    .reset_index()
)

yearly_crime
```

```python
state_crime = (
    df.groupby("STATE/UT")["TOTAL IPC CRIMES"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

state_crime
```

```python
scatter_data = df[["MURDER", "THEFT"]].dropna()

scatter_data.head()
```

```python
hist_data = df["TOTAL IPC CRIMES"].dropna()

hist_data.head()
```

```python
violin_data = df[
    ["STATE/UT", "TOTAL IPC CRIMES"]
].dropna()

violin_data.head()
```

```python
dashboard_visuals = {
    "yearly_crime": yearly_crime.to_dict(orient="records"),

    "state_crime": state_crime.to_dict(orient="records"),

    "scatter_data": scatter_data.to_dict(orient="records"),

    "hist_data": hist_data.tolist(),

    "violin_data": violin_data.to_dict(orient="records"),
}
```

```python
with open(
    "../frontend/src/data/dashboard_visuals.json",
    "w"
) as f:
    json.dump(dashboard_visuals, f)

print("dashboard_visuals.json created successfully")
```
