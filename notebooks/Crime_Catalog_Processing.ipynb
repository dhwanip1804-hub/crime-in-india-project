```python
import pandas as pd
import json
```

```python
df = pd.read_csv(
    "01_District_wise_crimes_committed_IPC_2001_2012.csv"
)

print("Dataset Loaded Successfully")
print(df.shape)
```

```python
ignore_columns = [
    "STATE/UT",
    "DISTRICT",
    "YEAR"
]

crime_columns = [
    col for col in df.columns
    if col not in ignore_columns
]

print("Crime Categories Found:")
print(crime_columns)
```

```python
catalog_data = []

ignore_columns = [
    "STATE/UT",
    "DISTRICT",
    "YEAR",
    "TOTAL IPC CRIMES"
]

crime_columns = [
    col for col in df.columns
    if col not in ignore_columns
]

years = f"{df['YEAR'].min()} - {df['YEAR'].max()}"

total_states = df["STATE/UT"].nunique()

for crime in crime_columns:

    total_cases = int(df[crime].sum())

    if total_cases <= 0:
        continue

    # Top state for this crime
    top_state = (
        df.groupby("STATE/UT")[crime]
        .sum()
        .sort_values(ascending=False)
        .index[0]
    )

    # Severity Logic
    if total_cases > 1000000:
        severity = "Critical Severity"

    elif total_cases > 500000:
        severity = "High Severity"

    elif total_cases > 100000:
        severity = "Medium Severity"

    else:
        severity = "Moderate Severity"

    # Related crimes
    related = [
        c.replace("_", " ").title()
        for c in crime_columns
        if c != crime
    ][:3]

    catalog_data.append({

        "crime":
            crime.title(),

        "type":
            "IPC Crime",

        "severity":
            severity,

        "states":
            f"{total_states} States",

        "years":
            years,

        "total_cases":
            total_cases,

        "top_state":
            top_state.title(),

        "insight":
            f"{crime.title()} recorded "
            f"{total_cases:,} reported cases "
            f"across the NCRB district dataset.",

        "related":
            related
    })

# Sort by highest crime count
# Keep only top 15 crimes
catalog_data = sorted(
    catalog_data,
    key=lambda x: x["total_cases"],
    reverse=True
)[:15]

print("Reduced Catalog Size")
print(len(catalog_data))
```

```python
import json

with open("catalog_data.json", "w") as file:
    json.dump(catalog_data, file, indent=2)

print("catalog_data.json exported successfully")
```
