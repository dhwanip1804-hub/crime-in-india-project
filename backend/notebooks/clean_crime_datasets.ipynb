```python
import pandas as pd
import numpy as np
import os
```

```python
def clean_dataset(input_path, output_name):

    print(f"\nCleaning: {output_name}")

    df = pd.read_csv(input_path)

    # STANDARDIZE COLUMN NAMES
    df.columns = (
        df.columns
        .str.strip()
        .str.upper()
        .str.replace("_", " ")
    )

    # REMOVE DUPLICATES
    df = df.drop_duplicates()

    # CONVERT POSSIBLE NUMERIC COLUMNS
    for col in df.columns:

        try:
            df[col] = pd.to_numeric(df[col])
        except:
            pass

    # REMOVE FULLY EMPTY ROWS
    df = df.dropna(how="all")

    # FILL REMAINING NULLS
    numeric_cols = df.select_dtypes(include=np.number).columns

    for col in numeric_cols:
        df[col] = df[col].fillna(df[col].median())

    text_cols = df.select_dtypes(include="object").columns

    for col in text_cols:
        df[col] = (
            df[col]
            .astype(str)
            .str.strip()
        )

    # REMOVE UNNAMED COLUMNS
    df = df.loc[:, ~df.columns.str.contains("^UNNAMED")]

    # SAVE CLEANED DATASET
    save_folder = "../backend/clean_datasets"

    os.makedirs(save_folder, exist_ok=True)

    save_path = f"{save_folder}/{output_name}"

    df.to_csv(save_path, index=False)

    print("Saved:", save_path)

    print("Shape:", df.shape)

    print("Columns:")
    print(df.columns.tolist())

    return df
```

```python
import pandas as pd
import os

# ====================================================
# FOLDERS
# ====================================================

INPUT_FOLDER = "raw_datasets"
OUTPUT_FOLDER = "clean_datasets"

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

# ====================================================
# DATASET CONFIGS
# ====================================================

datasets = [

    {
        "input": "children.csv",
        "output": "children_standardized.csv",

        "rename": {

            "BOYS 7-12 YEARS": "MURDER",
            "GIRLS 7-12 YEARS": "RAPE",
            "BOYS 12-16 YEARS": "KIDNAPPING",
            "BOYS 16-18 YEARS": "THEFT"

        }
    },

    {
        "input": "women.csv",
        "output": "women_standardized.csv",

        "rename": {

            "KIDNAPPING AND ABDUCTION": "KIDNAPPING"

        }
    },

    {
        "input": "ipc.csv",
        "output": "ipc_standardized.csv",

        "rename": {

            "KIDNAPPING & ABDUCTION": "KIDNAPPING"

        }
    },

    {
        "input": "sc.csv",
        "output": "sc_standardized.csv",

        "rename": {

            "KIDNAPPING ABDUCTION": "KIDNAPPING",
            "ROBBERY": "THEFT"

        }
    },

    {
        "input": "st.csv",
        "output": "st_standardized.csv",

        "rename": {

            "KIDNAPPING AND ABDUCTION": "KIDNAPPING",
            "ROBBERY": "THEFT"

        }
    }

]

# ====================================================
# PROCESS
# ====================================================

for dataset in datasets:

    input_path = os.path.join(
        INPUT_FOLDER,
        dataset["input"]
    )

    output_path = os.path.join(
        OUTPUT_FOLDER,
        dataset["output"]
    )

    df = pd.read_csv(input_path)

    # RENAME COLUMNS

    df = df.rename(
        columns=dataset["rename"]
    )

    # SAVE FULL DATASET
    # KEEP ALL COLUMNS

    df.to_csv(
        output_path,
        index=False
    )

    print(
        f"{dataset['output']} saved successfully"
    )

    print(df.columns.tolist())
    print("------------------------------------------------")
```
