```python
import pandas as pd
import os

OUTPUT_FOLDER = "clean_datasets"

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

# =========================
# IPC DATASET
# =========================

df = pd.read_csv("cleaned_ipc.csv")

df = df.rename(columns={

    "KIDNAPPING & ABDUCTION": "KIDNAPPING"

})

columns = [
    "STATE/UT",
    "DISTRICT",
    "MURDER",
    "RAPE",
    "KIDNAPPING",
    "THEFT"
]

df = df[columns]

df.to_csv(
    f"{OUTPUT_FOLDER}/ipc_standardized.csv",
    index=False
)

print("IPC standardized saved")


# =========================
# WOMEN DATASET
# =========================

df = pd.read_csv("cleaned_women.csv")

df = df.rename(columns={

    "KIDNAPPING AND ABDUCTION":
        "KIDNAPPING"

})

# WOMEN DATASET HAS NO THEFT
# CREATE SYNTHETIC COLUMN

df["THEFT"] = 0

columns = [
    "STATE/UT",
    "DISTRICT",
    "MURDER",
    "RAPE",
    "KIDNAPPING",
    "THEFT"
]

# WOMEN DATASET HAS NO MURDER
if "MURDER" not in df.columns:
    df["MURDER"] = 0

df = df[columns]

df.to_csv(
    f"{OUTPUT_FOLDER}/women_standardized.csv",
    index=False
)

print("Women standardized saved")


# =========================
# CHILDREN DATASET
# =========================

df = pd.read_csv("cleaned_children.csv")

df = df.rename(columns={

    "KIDNAPPING AND ABDUCTION":
        "KIDNAPPING"

})

df["THEFT"] = 0

columns = [
    "STATE/UT",
    "DISTRICT",
    "MURDER",
    "RAPE",
    "KIDNAPPING",
    "THEFT"
]

df = df[columns]

df.to_csv(
    f"{OUTPUT_FOLDER}/children_standardized.csv",
    index=False
)

print("Children standardized saved")


# =========================
# SC DATASET
# =========================

df = pd.read_csv("cleaned_sc.csv")

print(df.columns.tolist())

# AUTO RENAME POSSIBLE VARIANTS

rename_map = {}

for col in df.columns:

    if "KIDNAPPING" in col.upper():
        rename_map[col] = "KIDNAPPING"

    elif "MURDER" in col.upper():
        rename_map[col] = "MURDER"

    elif "RAPE" in col.upper():
        rename_map[col] = "RAPE"

df = df.rename(columns=rename_map)

# ADD MISSING COLUMNS

required_columns = [

    "STATE/UT",
    "DISTRICT",
    "MURDER",
    "RAPE",
    "KIDNAPPING",
    "THEFT"

]

for col in required_columns:

    if col not in df.columns:
        df[col] = 0

df = df[required_columns]

df.to_csv(
    f"{OUTPUT_FOLDER}/sc_standardized.csv",
    index=False
)

print("SC standardized saved")


# =========================
# ST DATASET
# =========================

df = pd.read_csv("cleaned_st.csv")

print(df.columns.tolist())

rename_map = {}

for col in df.columns:

    upper_col = col.upper()

    if "KIDNAPPING" in upper_col:
        rename_map[col] = "KIDNAPPING"

    elif "MURDER" in upper_col:
        rename_map[col] = "MURDER"

    elif "RAPE" in upper_col:
        rename_map[col] = "RAPE"

df = df.rename(columns=rename_map)

required_columns = [

    "STATE/UT",
    "DISTRICT",
    "MURDER",
    "RAPE",
    "KIDNAPPING",
    "THEFT"

]

# CREATE MISSING COLUMNS

for col in required_columns:

    if col not in df.columns:
        df[col] = 0

df = df[required_columns]

df.to_csv(
    f"{OUTPUT_FOLDER}/st_standardized.csv",
    index=False
)

print("ST standardized saved")
```
