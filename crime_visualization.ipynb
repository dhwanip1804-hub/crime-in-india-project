```python
import pandas as pd
import json
```

```python
df = pd.read_csv("01_District_wise_crimes_committed_IPC_2001_2012.csv")

df.head()
```

```python
timeline = []

for year in sorted(df["YEAR"].unique()):

    year_data = df[df["YEAR"] == year]

    murder = year_data["MURDER"].sum()
    rape = year_data["RAPE"].sum()
    theft = year_data["THEFT"].sum()
    robbery = year_data["ROBBERY"].sum()

    timeline.append({
        "YEAR": int(year),
        "MURDER": int(murder),
        "RAPE": int(rape),
        "THEFT": int(theft),
        "ROBBERY": int(robbery)
    })
```

```python
with open("visualization.json", "w") as file:
    json.dump(timeline, file, indent=4)

print("visualization.json created successfully")
```
